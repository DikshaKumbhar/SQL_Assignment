```python
# Question 1 : Show employees working in either the ‘IT’ or ‘HR’ departments.
df_it_hr = pd.read_sql("SELECT * FROM Employees WHERE Department IN ('IT','HR')", conn)
df_it_hr
```

```python
# Question 2: Employees in Sales, IT, or Finance departments
df_sales_it_finance = pd.read_sql(
    "SELECT * FROM Employees WHERE Department IN ('Sales', 'IT', 'Finance')",
    conn
)
df_sales_it_finance
```

```python
# Question 3: Employees with salary between 50,000 and 70,000
df_salary_range = pd.read_sql(
    "SELECT * FROM Employees WHERE Salary BETWEEN 50000 AND 70000",
    conn
)
df_salary_range
```

```python
# Question 4: Employees whose names start with 'A'
df_name_a = pd.read_sql(
    "SELECT * FROM Employees WHERE EmpName LIKE 'A%'",
    conn
)
df_name_a
```

```python
# Question 5: Employees whose names contain 'an'
df_name_an = pd.read_sql(
    "SELECT * FROM Employees WHERE EmpName LIKE '%an%'",
    conn
)
df_name_an
```

```python
# Question 6: Employees from Delhi or Mumbai with salary > 55000
df_delhi_mumbai = pd.read_sql(
    "SELECT * FROM Employees WHERE (City IN ('Delhi','Mumbai')) AND Salary > 55000",
    conn
)
df_delhi_mumbai
```

```python
# Question 7: All employees except those from HR department
df_not_hr = pd.read_sql(
    "SELECT * FROM Employees WHERE Department <> 'HR'",
    conn
)
df_not_hr
```

```python
# Question 8: Employees hired between 2019 and 2022, ordered by HireDate
df_hiredate = pd.read_sql(
    "SELECT * FROM Employees WHERE HireDate BETWEEN '2019-01-01' AND '2022-12-31' ORDER BY HireDate ASC",
    conn
)
df_hiredate
```
